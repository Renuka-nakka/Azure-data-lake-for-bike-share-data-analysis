
## Overview

This notebook will show you how to create and query a table or DataFrame that you uploaded to DBFS. [DBFS](https://docs.databricks.com/user-guide/dbfs-databricks-file-system.html) is a Databricks File System that allows you to store data for querying inside of Databricks. This notebook assumes that you have a file already inside of DBFS that you would like to read from.

This notebook is written in **Python** so the default cell type is Python. However, you can use different languages by using the `%LANGUAGE` syntax. Python, Scala, SQL, and R are all supported.

```python
dim_rider = spark.sql("""
SELECT
    rider_id,
    first,
    last,
    address,
    birthday,
    account_start_date,
    account_end_date,
    is_member
FROM divvy.rider_bronze
""")
```

```python
dim_station = spark.sql("""
SELECT DISTINCT
    station_id,
    name,
    latitude,
    longitude
FROM divvy.station_bronze
""")
```

```python
dim_station.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("divvy.dim_station")
```

```python
from pyspark.sql.functions import *

# Dates from trip started_at
trip_start_dates = spark.sql("""
SELECT
    to_date(
        to_timestamp(started_at, 'dd-MM-yyyy HH:mm')
    ) AS date
FROM divvy.trip_bronze
""")

# Dates from trip ended_at
trip_end_dates = spark.sql("""
SELECT
    to_date(
        to_timestamp(ended_at, 'dd-MM-yyyy HH:mm')
    ) AS date
FROM divvy.trip_bronze
""")

# Dates from payments
payment_dates = spark.sql("""
SELECT
    to_date(date, 'dd-MM-yyyy') AS date
FROM divvy.payment_bronze
""")

# Combine ALL dates
all_dates = trip_start_dates.union(trip_end_dates) \
.union(payment_dates) \
.distinct()

# Create final date dimension
dim_date = all_dates.select(
    col("date"),
    year("date").alias("year"),
    quarter("date").alias("quarter"),
    month("date").alias("month"),
    dayofmonth("date").alias("day"),
    dayofweek("date").alias("day_of_week")
).distinct()

display(dim_date)
```

```python
dim_date.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("divvy.dim_date")
```

```python
from pyspark.sql.functions import *

fact_trip = spark.sql("""
SELECT
    t.ride_id,
    t.rider AS rider_id,
    t.start_station_id,
    t.end_station_id,

    to_timestamp(t.started_at, 'dd-MM-yyyy HH:mm') AS started_at,
    to_timestamp(t.ended_at, 'dd-MM-yyyy HH:mm') AS ended_at,

    CAST(
        floor(
            datediff(
                to_timestamp(t.started_at, 'dd-MM-yyyy HH:mm'),
                to_date(r.birthday, 'dd-MM-yyyy')
            ) / 365.25
        ) AS INT
    ) AS rider_age

FROM divvy.trip_bronze t
JOIN divvy.rider_bronze r
ON t.rider = r.rider_id
""")
```

```python
fact_trip = fact_trip.withColumn(
    "trip_duration_minutes",
    (
        unix_timestamp(col("ended_at")) -
        unix_timestamp(col("started_at"))
    ) / 60
)
```

```python
fact_trip.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("divvy.fact_trip")
```
